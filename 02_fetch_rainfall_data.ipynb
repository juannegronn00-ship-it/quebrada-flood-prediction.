```python
import requests
import h5py
import numpy as np
import tempfile
import os

def get_daily_rainfall(lat, lon, date_str, username, password):
    """
    Get daily rainfall in mm for a specific lat/lon and date
    date_str format: YYYYMMDD
    """
    year = date_str[:4]
    month = date_str[4:6]
    
    filename = f"3B-DAY.MS.MRG.3IMERG.{date_str}-S000000-E235959.V07B.nc4"
    url = f"https://gpm1.gesdisc.eosdis.nasa.gov/data/GPM_L3/GPM_3IMERGDF.07/{year}/{month}/{filename}"
    
    response = requests.get(url, auth=(username, password))
    
    if response.status_code != 200:
        print(f"Failed: {response.status_code}")
        return None
    
    # Save to temp file and read with h5py
    with tempfile.NamedTemporaryFile(suffix='.nc4', delete=False) as f:
        f.write(response.content)
        tmp_path = f.name
    
    with h5py.File(tmp_path, 'r') as f:
        rainfall = f['precipitationCal'][:]
        lats = f['lat'][:]
        lons = f['lon'][:]
    
    os.unlink(tmp_path)
    
    # Find closest grid point
    lat_idx = np.argmin(np.abs(lats - lat))
    lon_idx = np.argmin(np.abs(lons - lon))
    
    return float(rainfall[0, lon_idx, lat_idx])

# Test with first flood event
rain = get_daily_rainfall(18.3988, -66.0833, "20150520", "juan360", "@8Yd9fmQP/^^&ne")
print(f"Rainfall on flood day: {rain} mm")
```

```python
import time

def get_rainfall_for_event(lat, lon, date_str, username, password, days_back=3):
    """Get rainfall for multiple days leading up to a flood event"""
    from datetime import datetime, timedelta
    
    results = {}
    date = datetime.strptime(date_str, '%Y%m%d')
    
    for i in range(days_back + 1):
        target_date = date - timedelta(days=i)
        target_str = target_date.strftime('%Y%m%d')
        rain = get_daily_rainfall(lat, lon, target_str, username, password)
        results[f'rain_day_minus_{i}'] = rain
        time.sleep(0.5)  # be polite to NASA servers
    
    return results

# Test with first event
event = df.iloc[0]
date_str = datetime.strptime(event['BEGIN_DATE'], '%m/%d/%Y').strftime('%Y%m%d')

print(f"Fetching 4 days of rainfall for {event['CZ_NAME_STR']} on {event['BEGIN_DATE']}...")
rainfall_data = get_rainfall_for_event(
    event['BEGIN_LAT'], event['BEGIN_LON'], 
    date_str, "juan360", "@8Yd9fmQP/^^&ne"
)

print("\nRainfall leading up to flood:")
for key, val in rainfall_data.items():
    print(f"  {key}: {val:.2f} mm")
```

```python
import srtm
import pandas as pd
import numpy as np

df = pd.read_csv('/Users/juannegron/quebrada-flood-prediction./data/processed/flood_rainfall_data.csv')

elevation_data = srtm.get_data()

def get_slope(lat, lon, delta=0.001):
    """Estimate slope in degrees from neighboring SRTM points"""
    e_center = elevation_data.get_elevation(lat, lon) or 0
    e_north  = elevation_data.get_elevation(lat + delta, lon) or 0
    e_south  = elevation_data.get_elevation(lat - delta, lon) or 0
    e_east   = elevation_data.get_elevation(lat, lon + delta) or 0
    e_west   = elevation_data.get_elevation(lat, lon - delta) or 0
    
    # delta in degrees → meters (approx)
    dist = delta * 111000
    
    dz_dx = (e_east - e_west) / (2 * dist)
    dz_dy = (e_north - e_south) / (2 * dist)
    
    slope_deg = np.degrees(np.arctan(np.sqrt(dz_dx**2 + dz_dy**2)))
    return round(slope_deg, 2)

slopes = []
for i, row in df.iterrows():
    slope = get_slope(row['lat'], row['lon'])
    slopes.append(slope)
    if (i + 1) % 50 == 0:
        print(f"{i+1}/500 done")

df['slope_deg'] = slopes

print(f"\nSlope range: {df['slope_deg'].min()}° to {df['slope_deg'].max()}°")
print(df[['location', 'elevation_m', 'slope_deg']].head(10))

df.to_csv('/Users/juannegron/quebrada-flood-prediction./data/processed/flood_rainfall_data.csv', index=False)
print("\nSaved with slope column added.")
```

```python
def is_karst_zone(lat, lon, soil_compname):
    # Method 1: Known karst soil series in Puerto Rico
    karst_soils = ['Mucara', 'Colinas', 'Montegrande', 'Limestone']
    soil_karst = any(k.lower() in soil_compname.lower() for k in karst_soils)
    
    # Method 2: Northern Karst Belt coordinates (USGS Puerto Rico karst mapping)
    # Northern belt runs along north coast roughly lat 18.1-18.52, lon -67.3 to -65.8
    geo_karst = (18.1 <= lat <= 18.52) and (-67.3 <= lon <= -65.8)
    
    # Southern karst (Ponce/Guanica area)
    south_karst = (17.9 <= lat <= 18.1) and (-67.0 <= lon <= -66.4)
    
    return 1 if (soil_karst or geo_karst or south_karst) else 0

df = pd.read_csv('/Users/juannegron/quebrada-flood-prediction./data/processed/flood_rainfall_data.csv')

df['is_karst'] = df.apply(
    lambda row: is_karst_zone(row['lat'], row['lon'], str(row['soil_compname'])), 
    axis=1
)

print(f"Karst events: {df['is_karst'].sum()} / {len(df)}")
print(f"Non-karst events: {(df['is_karst']==0).sum()} / {len(df)}")

df.to_csv('/Users/juannegron/quebrada-flood-prediction./data/processed/flood_rainfall_data.csv', index=False)
print("\nFinal dataset columns:")
print(list(df.columns))
print(f"\nShape: {df.shape}")
```
