```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the flood events
df = pd.read_csv('../data/labels/storm_data_search_results.csv')

# Basic overview
print("Total flood events:", len(df))
print("Deaths:", df['DEATHS_DIRECT'].sum())
print("Date range:", df['BEGIN_DATE'].min(), "to", df['BEGIN_DATE'].max())
print("\nTop 10 most affected areas:")
print(df['CZ_NAME_STR'].value_counts().head(10))
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/Users/juannegron/quebrada-flood-prediction./data/labels/storm_data_search_results.csv')

print("Total flood events:", len(df))
print("Deaths:", df['DEATHS_DIRECT'].sum())
print("Date range:", df['BEGIN_DATE'].min(), "to", df['BEGIN_DATE'].max())
print("\nTop 10 most affected areas:")
print(df['CZ_NAME_STR'].value_counts().head(10))
```

```python
import matplotlib.pyplot as plt

# Plot flood locations
plt.figure(figsize=(10, 8))
plt.scatter(df['BEGIN_LON'], df['BEGIN_LAT'], 
            c=df['DEATHS_DIRECT'], cmap='Reds', 
            alpha=0.6, s=50)
plt.colorbar(label='Deaths')
plt.title('Puerto Rico Flash Flood Events 2015-2024')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.grid(True)
plt.show()
```
